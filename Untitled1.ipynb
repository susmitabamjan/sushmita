```python
import cv2

# 1. Load the face detector
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

# 2. Check if the face detector loaded
if face_cascade.empty():
    print("Error: Face detector could not be loaded.")
else:
    print("Face detector loaded successfully!")

    # 3. Open the webcam
    cap = cv2.VideoCapture(0)

    if not cap.isOpened():
        print("Error: Could not open webcam.")

    else:
        try:
            # 4. Start face detection
            while True:

                # Read webcam frame
                ret, frame = cap.read()

                if not ret:
                    print("Error: Could not read frame.")
                    break

                # 5. Convert frame to grayscale
                gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

                # 6. Detect faces
                faces = face_cascade.detectMultiScale(
                    gray,
                    scaleFactor=1.1,
                    minNeighbors=5,
                    minSize=(30, 30)
                )

                # 7. Draw rectangle around faces
                for (x, y, w, h) in faces:
                    cv2.rectangle(
                        frame,
                        (x, y),
                        (x + w, y + h),
                        (255, 0, 0),
                        2
                    )

                # 8. Show webcam
                cv2.imshow("Face Detection", frame)

                # 9. Press Q to close
                if cv2.waitKey(1) & 0xFF == ord("q"):
                    break

        finally:
            # 10. Release the webcam
            cap.release()

            # 11. Close all OpenCV windows
            cv2.destroyAllWindows()

            print("Camera closed.")
            print("All OpenCV windows closed.")
```
